# Time-Wasters on Social Media
This dataset contains user behavior and demographic data related to social media usage . It captures various aspects of user activity, including time spent, engagement metrics, and platforms used. Additionally, it includes demographic and contextual details such as age, gender, income, device type, and connection type.

The dataset provides a valuable resource for analyzing social media consumption patterns, factors influencing productivity loss, and behavioral trends across different demographic segments.

```python
data=pd.read_csv('Time-Wasters on Social Media.csv')
```

## Key Features


* UserID: Unique identifier assigned to each user.
* Age: The user's age.
* Gender: The user's gender (e.g., male, female, non-binary).
* Location: Geographic location of the user.
* Income: The user's income level.
* Debt: Amount of debt the user has.
* Owns Property: Indicates whether the user owns property.
* Profession: The user's occupation or job.
* Demographics: Statistical data about the user (e.g., age, gender, income).
* Platform: The platform the user is using (e.g., website, mobile app).
* Total Time Spent: The total time the user spends on the platform.
* Number of Sessions: The number of times the user has logged into the platform.
* Video ID: Unique identifier for a video.
* Video Category: The category or genre of the video.
* Video Length: Duration of the video.
* Engagement: User interaction with the video (e.g., likes, comments, shares).
* Importance Score: A score indicating how important the video is to the user.
* Time Spent On Video: The amount of time the user spends watching a video.
* Number of Videos Watched: The total number of videos watched by the user.
* Scroll Rate: The rate at which the user scrolls through content.
* Frequency: How often the user engages with the platform.
* Productivity Loss: The impact of platform usage on the user's productivity.
* Satisfaction: The user's satisfaction level with the platform or content.
* Watch Reason: The reason why the user is watching a video (e.g., entertainment, education).
* Device Type: The type of device the user is using (e.g., smartphone, tablet, desktop).
* OS: The operating system of the user's device (e.g., iOS, Android, Windows).
* Watch Time: The time of day when the user watches videos.
* Self Control: The user's ability to control their usage of the platform.
* Addiction Level: The user's level of dependency on the platform.
* Current Activity: What the user is doing while watching the video.
* Connection Type: The type of internet connection the user has (e.g., Wi-Fi, cellular).

### Box plot

```python
plt.figure(figsize=(10,5))
df.boxplot()
plt.xticks(rotation=90)
plt.savefig('box plot.png',dpi=300)
plt.show()
```

### Pie plot

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%',shadow=True,startangle=60,colors=['darkred','pink','red'],explode=(0,0,0.1))
plt.title('Gender wise classifiction',fontsize=15,fontname='Times New Roman',fontweight='bold')
plt.savefig('gender wise classification.png',dpi=300)
plt.show()
```

```python

df['Age'].value_counts().head(10).plot(kind='bar',color='green',edgecolor='k')
plt.title('Age wise classification',fontsize=15,fontname='Times New Roman',fontweight='bold')
plt.savefig('age wise classification.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.title('Satisfactory levels by gender',color='black',fontsize=18,fontweight='bold')
sns.countplot(x='Satisfaction',data=df,hue='Gender',palette=['red','blue','green'])
plt.savefig('Satisfaction.png',dpi=300)
plt.show()
```

```python
dg=df[['Age','Time Spent On Video','Satisfaction','Self Control','Addiction Level','ProductivityLoss']].corr()
dg
```

```python
plt.figure(figsize=(10,5))
plt.title('Correlation Heatmap of Social media time-Waters',color='black',fontsize=18,fontname='Times New Roman',fontweight='bold')
numer=df.select_dtypes('number')
sns.heatmap(dg,annot=True,cmap='pink')
plt.savefig('heat map.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10,5))
df.groupby('Location')['Addiction Level'].sum().plot(kind='barh',color='darkred')
plt.title('Addiction Level in Location base',fontsize=18,fontname='Times New Roman',fontweight='bold',c='k')
plt.savefig('addiction level location .png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Gender',y='Addiction Level',data=df,hue='Watch Reason',palette='Reds')
plt.title('Analysis Based on Addiction level & Reason',fontsize=18,fontname='Times New Roman',fontweight='bold')
plt.xlabel('Demographics',color='darkred',fontsize=11,fontname='Times New Roman',fontweight='bold')
plt.ylabel('Addiction Level',color='darkred',fontsize=11,fontname='Times New Roman',fontweight='bold')
plt.savefig('addiction level reasons.png',dpi=300)
plt.show()
```

```python
df['ConnectionType'].value_counts().plot(kind='pie',autopct='%1.0f%%',shadow=True,colors=['darkred','lightgreen'])
plt.title('connection Type',color='black',fontsize=18,fontweight='bold')
plt.savefig('connection type.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['OS'], bins=10, kde=True, color='darkred')
plt.title('Most used OS ', fontsize=18, fontweight='bold')
plt.xlabel('OS',color='r',fontsize=16)
plt.ylabel('Frequency',color='r',fontsize=16)
plt.savefig('os.png',dpi=300)
plt.show()
```

## Findings

* Males are to spend more time on social media than females and show higher levels of addiction.

* The age group 43 is the most frequent or highest count in the dataset, followed closely by ages 50 and 45..

* Students and waiting staff are the heaviest users of social media, whereas teaching professionals have the lowest usage rates.

* YouTube is the most widely used platform for time-wasting activities across.

* Most used network is mobile network ,wifi users are very few.
